```python

import pandas as pd


def load_data(file_path: str) -> pd.DataFrame:
    """
    Загружает данные из CSV-файла.
    """
    data = pd.read_csv(file_path)

    return data


def preprocess_data(data: pd.DataFrame) -> pd.DataFrame:
    """
    Выполняет предварительную обработку временного ряда.
    """

    # Создаём копию исходных данных
    data = data.copy()

    # Преобразуем Timestamp в формат даты и времени
    data["Timestamp"] = pd.to_datetime(
        data["Timestamp"],
        errors="coerce"
    )

    # Удаляем строки с некорректными временными отметками
    data = data.dropna(subset=["Timestamp"])

    # Сортируем данные по времени
    data = data.sort_values(by="Timestamp")

    # Удаляем полные дубликаты
    data = data.drop_duplicates()

    # Проверяем пропущенные значения
    print("Пропущенные значения до обработки:")
    print(data.isnull().sum())

    # Удаляем строки с пропусками
    data = data.dropna()

    # Сбрасываем индексы
    data = data.reset_index(drop=True)

    print("\nРазмер данных после обработки:")
    print(data.shape)

    print("\nПропущенные значения после обработки:")
    print(data.isnull().sum())

    return data
```
